#  Seaborn – Class 5 & 6: Matrix/Grid Plots + Styling


## Why Matrix / Grid Plots?

Matrix or grid-style plots are perfect for:
- Showing correlations between variables  
- Visualizing pivot tables  
- Discovering patterns in large datasets


### Heatmaps (heatmap())

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset("tips")
tips.head()
```

```python
tips.corr(numeric_only=True)
```

```python
sns.heatmap(tips.corr(numeric_only=True))
```

```python
sns.heatmap(tips.corr(numeric_only=True),annot=True) #use only in numerical value
```

### Cluster Map (clustermap())

```python
sns.clustermap(tips.corr(numeric_only=True))
```

```python
sns.clustermap(tips.corr(numeric_only=True),annot=True)
```

### Styling & Themes

```python
sns.set_style("dark")
sns.set_palette('Set1')
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'],s=50)
plt.grid(True)
```
