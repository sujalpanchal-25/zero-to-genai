# **Questions**
1) Are most customer bills small or large? (Histogram)

2) Which gender visits the restaurant more: Male or Female? (Count Plot)

3) Which day has the highest average total bill? (Bar Plot)

4) Which day has the greatest variation (spread) in total bills? (Box Plot)

5) On which day are customer bills most concentrated? (Violin Plot)

6) Are there any unusually high or low bills? If yes, on which day are they observed? (Strip Plot)

7) Does a higher total bill usually result in a higher tip? (Scatter Plot)

8) What relationships exist among all the numerical variables in the dataset? (Pair Plot)

9) Which numerical variables are most strongly correlated? (Heatmap)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset("tips")
tips.head()
```

```python
sns.histplot(x=tips['total_bill'],bins=28)
```

```python
sns.countplot(x=tips['sex'],hue=tips['sex'])
```

```python
sns.barplot(x=tips['day'],y=tips['total_bill'])
```

```python
sns.violinplot(x=tips['day'],y=tips['total_bill'])
```

```python
sns.boxenplot(x = tips['day'],y=tips['total_bill'])
```

```python
sns.stripplot(x = tips['size'],y=tips['total_bill'])
```

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'])
```

```python
sns.pairplot(tips,)
```

```python
corr = tips.select_dtypes("number").corr()
sns.heatmap(corr, annot=True)
```

# **ANWER CODE**


```python
plt.figure(figsize=(8,5))
sns.histplot(data=tips, x="total_bill", bins=20, kde=True, color="skyblue")

plt.title("Distribution of Total Bills")
plt.xlabel("Total Bill")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.countplot(data=tips, x="sex", palette="Set2")

plt.title("Male vs Female Customers")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.barplot(data=tips, x="day", y="total_bill", palette="viridis")

plt.title("Average Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Bill")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.boxplot(data=tips, x="day", y="total_bill", palette="Pastel1")

plt.title("Bill Distribution by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.violinplot(data=tips, x="day", y="total_bill", palette="Set3")

plt.title("Distribution of Bills by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.stripplot(data=tips, x="day", y="total_bill", jitter=True, palette="deep")

plt.title("Individual Bills by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex", palette="Set1")

plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
sns.pairplot(tips, hue="sex")

plt.show()
```

```python
corr = tips.select_dtypes(include="number").corr()

plt.figure(figsize=(6,5))
sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```
